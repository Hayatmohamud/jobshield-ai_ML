1) Ka saar duplicates.

```python
import pandas as pd
df = pd.read_csv("../dataset/raw/fake_job_postings.csv")

df_clean = df.copy()

print(df_clean.duplicated().sum())

df_clean = df_clean.drop_duplicates()

print(df_clean.shape)
```

2) no prediction

```python
df_clean = df_clean.drop(columns=["job_id"])
```

3) no missing value

```python
df_clean.isnull().sum()
```

4) new features

```python
df_clean["text"] = (
    df_clean["title"] + " " +
    df_clean["company_profile"] + " " +
    df_clean["description"] + " " +
    df_clean["requirements"] + " " +
    df_clean["benefits"]
)
```

```python
df_clean[["text", "fraudulent"]].head()
```

5) save cleaned Data

```python
from pathlib import Path

output_path = Path("../dataset/processed")

# Samee folder-ka haddii uusan jirin
output_path.mkdir(parents=True, exist_ok=True)

# Save cleaned dataset
df_clean.to_csv(
    output_path / "fake_job_postings_clean.csv",
    index=False
)

print("✅ Cleaned dataset saved successfully!")
```
